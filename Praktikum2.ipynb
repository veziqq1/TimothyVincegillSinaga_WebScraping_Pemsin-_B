```python
import requests
import pandas as pd
from io import StringIO

header = {"User-Agent": "Mozilla/5.0"}

# Web Scrape 1
url_worldometer = "https://www.worldometers.info/world-population/population-by-country/"
req_worldometer = requests.get(
    url_worldometer, headers=header, timeout=30
)
req_worldometer.raise_for_status()

html = req_worldometer.text

tabel_worldometer = pd.read_html(StringIO(html))
df_populasi = tabel_worldometer[0]

print("Data Scraping Worldometer:")
display(df_populasi.head())
```

```python
#WEB SCRAPE 2
url_scrape = "https://www.scrapethissite.com/pages/forms/"
req_scrape = requests.get(url_scrape, headers=header)

tabel_scrape = pd.read_html(StringIO(req_scrape.text))
df_hoki = tabel_scrape[0]

print("Data Scraping ScrapeThisSites:")
display(df_hoki.head)
```
